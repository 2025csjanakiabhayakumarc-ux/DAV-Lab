```python
import pandas as pd
import numpy as np
```

```python
arr = np.array([15, 25, 35, 45, 55, 65])
print("Array:", arr)
print("Mean:", arr.mean())
print("Reshaped 2D:\n", arr.reshape(6, 1))
print("Slice (index 1 to 4):", arr[1:5])
```

```python
d = {
    "Name": ["Anita", "Vikram", "Sneha", "Farhan"],
    "Marks": [92, 74, 85, 63],
    "Branch": ["ISE", "CSE", "AI", "ECE"]
}
data = pd.DataFrame(d)
print(data)
```

```python
print("First 2 Rows\n", data.iloc[:2])
```

```python
print("Marks Column:\n", data["Marks"])
```

```python
print("Rows where Marks > 70:\n", data[data["Marks"] > 70])
```
