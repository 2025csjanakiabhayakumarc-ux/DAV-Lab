## Filtering Data
we select customers whose age is less than 20

```python
df[df["Age"]<20]
```

## Sorting the Dataset
customers are sorted according to their Spending Score in ascending order

```python
df.sort_values("Spending Score (1-100)")
```

## Renaming Columns
the `Age` column is renamed to `Age in years` to make its meaning clearer

```python
df.rename(columns={"Age":"Age in years"})
```

## Creating a New Column

Annual Income is given in thousands of dollars (`k$`)

```python
df["Annual Income ($)"] = df["Annual Income (k$)"] * 1000
```

### Viewing Data Except the Last 15 Rows

```python
df.head(-15)
```

### Groups by Gender, finds average spending score

```python
df.groupby("Gender")[["Annual Income (k$)", "Spending Score (1-100)"]].mean()
```

## Removing an Unnecessary Column

The `Annual Income ($)` column was created temporarily for conversion purposes

```python
df = df.drop(columns=["Annual Income ($)"])
```
